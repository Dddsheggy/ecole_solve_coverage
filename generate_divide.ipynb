```python
import os
import numpy as np
```

```python
def generate_divide(task_num, uav_num, task_min, task_max, ins_num, rng, lp_dir):

    for ins in range(ins_num):
        filename = os.path.join(lp_dir, f'instance_{ins + 1}.lp')
        with open(filename, 'w') as file:
            # write objective function
            file.write("minimize\nOBJ:")
            coefs = rng.uniform(low=0, high=np.sqrt(25000*25000), size=task_num*uav_num).tolist()
            obj = ""
            for i in range(uav_num):
                obj += "".join([f" +{coefs[i * uav_num + j]} x_{i + 1}_{j + 1}" for j in range(task_num)])
            obj += '\n'
            file.write(obj)

            # write constraints
            file.write("\nsubject to\n")
            for i in range(task_num):
                cons_type1 = "".join([f"+x_{j + 1}_{i + 1} " for j in range(uav_num)]) + "= 1\n"
                file.write(cons_type1)


            for i in range(uav_num):
                cons_type2_max = "".join([f"+x_{i + 1}_{j + 1} " for j in range(task_num)]) + f"<= {task_max}\n"
                cons_type2_min = "".join([f"+x_{i + 1}_{j + 1} " for j in range(task_num)]) + f">= {task_min}\n"
                file.write(cons_type2_max)
                file.write(cons_type2_min)

            # # write variable bounds
            # file.write("\nbounds\n")
            # for i in range(uav_num):
            #     for j in range(task_num):
            #         bound = f"x_{i + 1}_{j + 1} <= 1\n"
            #         file.write(bound)
            
            # # write integer variables
            # file.write("\ngeneral\n")
            # general = "".join([f"x_{i + 1}_{j + 1} " for i in range(uav_num) for j in range(task_num)]) + "\n"
            # file.write(general)

            # write binary variables
            file.write("\nbinary\n")
            binary = "".join([f"x_{i + 1}_{j + 1} " for i in range(uav_num) for j in range(task_num)]) + "\n"
            file.write(binary)

            # write end
            file.write("\nend")
            
```

```python
task_num = 69
uav_num = 7
task_min = 6
task_max = 10
train_ins_num = 1
rng = np.random.RandomState(0)
# lp_dir = 'instance_divide/train/'
lp_dir = ''
# os.makedirs(lp_dir, exist_ok=False)

print(f'{uav_num} UAVs conduct {task_num} tasks, each UAV has {task_min}-{task_max} tasks')
print(f'generating {train_ins_num} training instances in {lp_dir}')
generate_divide(task_num, uav_num, task_min, task_max, train_ins_num, rng, lp_dir)
print('done')
```

```python
valid_ins_num = 1000
rng = np.random.RandomState(1)
lp_dir = 'instance_divide/valid/'
os.makedirs(lp_dir, exist_ok=False)

print(f'generating {valid_ins_num} validation instances in {lp_dir}')
generate_divide(task_num, uav_num, task_min, task_max, valid_ins_num, rng, lp_dir)
print('done')
```

```python
test_ins_num = 1000
rng = np.random.RandomState(2)
lp_dir = 'instance_divide/test/'
os.makedirs(lp_dir, exist_ok=False)

print(f'generating {test_ins_num} testing instances in {lp_dir}')
generate_divide(task_num, uav_num, task_min, task_max, test_ins_num, rng, lp_dir)
print('done')
```
