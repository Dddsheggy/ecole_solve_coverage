```python
import pickle
import gzip
import numpy as np
```

```python
sample_trial = 'samples/sample_1.pkl'

with gzip.open(sample_trial, 'rb') as f:
    d = pickle.load(f)

# print(d)

for data in d[2:]:
    print(data)
    print(len(data))
```

```python
sample_setcover = 'samples_setcover/sample_1.pkl'

with gzip.open(sample_setcover, 'rb') as f:
    d = pickle.load(f)

# print(d)

for data in d[2:]:
    print(data)
    print(len(data))
    cnt = 0
    
    for num in data:
        if np.isnan(num):
            cnt += 1
    print(cnt)
```

```python
import pyscipopt as sp

model = sp.Model()
filename = 'divide.lp'
model.readProblem(filename)
model.optimize()
vars = model.getVars()
for var in vars:
    value = model.getVal(var)
    if value:
        print(var)
```
