```python
import os
import numpy as np
```

```python
def generate_plan(task_num, max_dis, i_c, ins_num, rng, lp_dir, to_solve=True):
    for ins in range(ins_num):
        filename = os.path.join(lp_dir, f'instance_{ins + 1}.lp')
        with open(filename, 'w') as file:
            # confirm i_c
            if not to_solve:
                i_c = rng.randint(task_num)

            # write objective function
            file.write("minimize\nOBJ:")
            obj = ""
            for i in range(task_num):
                for j in range(task_num):
                    if (j != i) and (j != i_c):
                        coef = rng.uniform(low=0, high=max_dis)
                        obj += f"+{coef} x_{i + 1}_{j + 1} "
            obj += '\n'
            file.write(obj)

            # write constraints
            file.write("\nsubject to\n")
            for i in range(task_num):
                cons_type1_out = "".join([f"+x_{i + 1}_{j + 1} " for j in range(task_num) if j != i]) + "= 1\n"
                file.write(cons_type1_out)

            for i in range(task_num):
                cons_type1_in = "".join([f"+x_{j + 1}_{i + 1} " for j in range(task_num) if j != i]) + "= 1\n"
                file.write(cons_type1_in)

            for i in range(task_num):
                if i != i_c:
                    for j in range(task_num):
                        if (j != i) and (j != i_c):
                            cons_type2 = f"+{task_num} x_{i + 1}_{j + 1} +u_{i + 1} -u_{j + 1} <= {task_num - 1}\n"
                            file.write(cons_type2)

            # write variable bounds
            file.write("\nbounds\n")
            for i in range(task_num):
                for j in range(task_num):
                    bound = f"x_{i + 1}_{j + 1} <= 1\n"
                    file.write(bound)
            
            # write integer variables
            file.write("\ngeneral\n")
            general = "".join([f"x_{i + 1}_{j + 1} " for i in range(task_num) for j in range(task_num)]) + "\n"
            file.write(general)

            # write end
            file.write("\nend")
            
```

```python
task_num = 10
max_dis = np.sqrt(25000 * 25000)
i_c = 0

train_ins_num = 1
rng = np.random.RandomState(0)
lp_dir = 'instances_plan/'
os.makedirs(lp_dir, exist_ok=True)

to_solve = True

print(f'{task_num} tasks for every UAV')
print(f'generating {train_ins_num} training instances in {lp_dir}')
generate_plan(task_num, max_dis, i_c, train_ins_num, rng, lp_dir, to_solve)
print('done')
```
